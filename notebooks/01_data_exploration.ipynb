```python
import pandas as pd
```

```python
video_url = "https://www.youtube.com/watch?v=Oa-Ah1Xz290"

video_id = video_url.split("v=")[1]

print(video_id)
```

```python
from youtube_transcript_api import YouTubeTranscriptApi

ytt_api = YouTubeTranscriptApi()

transcript = ytt_api.fetch(video_id)

print(transcript)
```

```python
transcript_text = " ".join(snippet.text for snippet in transcript)

print(transcript_text)
```

```python
print("Number of characters:", len(transcript_text))
print("Number of words:", len(transcript_text.split()))
```

```python
# Look at the first 10 transcript snippets
for i, snippet in enumerate(transcript[:10]):
    print(f"Snippet {i}")
    print("Text:", snippet.text)
    print("Start:", snippet.start)
    print("Duration:", snippet.duration)
    print("-" * 50)
```

```python
# Basic statistics about the transcript snippets

num_snippets = len(transcript)

snippet_word_counts = [
    len(snippet.text.split())
    for snippet in transcript
]

snippet_char_counts = [
    len(snippet.text)
    for snippet in transcript
]

print("Number of snippets:", num_snippets)
print("Average words per snippet:", sum(snippet_word_counts) / num_snippets)
print("Average characters per snippet:", sum(snippet_char_counts) / num_snippets)
print("Shortest snippet (words):", min(snippet_word_counts))
print("Longest snippet (words):", max(snippet_word_counts))
```

```python
# Check how much consecutive transcript snippets overlap in time

overlap_count = 0

for i in range(len(transcript) - 1):
    current = transcript[i]
    next_snippet = transcript[i + 1]

    current_end = current.start + current.duration

    if next_snippet.start < current_end:
        overlap_count += 1

print("Overlapping consecutive snippets:", overlap_count)
```

```python
import re

# Count some basic punctuation marks in the transcript
periods = transcript_text.count(".")
question_marks = transcript_text.count("?")
exclamations = transcript_text.count("!")

print("Periods:", periods)
print("Question marks:", question_marks)
print("Exclamation marks:", exclamations)
```

```python
WINDOW_SIZE = 75

windows = []

current_words = []
window_start = None
window_end = None

for snippet in transcript:

    words = snippet.text.split()

    # Set the start time when a new window begins
    if window_start is None:
        window_start = snippet.start

    # Add the current snippet's words
    current_words.extend(words)

    # Keep updating the end time
    window_end = snippet.start + snippet.duration

    # Once we have enough words, create a window
    if len(current_words) >= WINDOW_SIZE:

        windows.append({
            "id": len(windows),
            "text": " ".join(current_words),
            "start": window_start,
            "end": window_end
        })

        # Reset for the next window
        current_words = []
        window_start = None
        window_end = None


# Add the final smaller window if anything remains
if current_words:

    windows.append({
        "id": len(windows),
        "text": " ".join(current_words),
        "start": window_start,
        "end": window_end
    })
```

```python
print("Number of windows:", len(windows))

print("\nFirst window:")
print("ID:", windows[0]["id"])
print("Start:", windows[0]["start"])
print("End:", windows[0]["end"])
print("Word count:", len(windows[0]["text"].split()))
print("Text:", windows[0]["text"])
```

```python
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
embedding = model.encode(windows[0]["text"])

print("Embedding shape:", embedding.shape)
print("First 10 values:", embedding[:10])
```

```python
import numpy as np

embedding = np.array(embedding)

print("Shape:", embedding.shape)
print("Data type:", embedding.dtype)
print("Memory:", embedding.nbytes / 1024, "KB")
```

```python
embeddings = model.encode(
    [window["text"] for window in windows],
    show_progress_bar=True
)

print("Embeddings shape:", embeddings.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarities = []

for i in range(len(embeddings) - 1):
    score = cosine_similarity(
        embeddings[i].reshape(1, -1),
        embeddings[i + 1].reshape(1, -1)
    )[0][0]

    similarities.append(score)

print("Number of similarity scores:", len(similarities))
print("First 10 scores:", similarities[:10])
```

```python
import numpy as np

similarities = np.array(similarities)

threshold = np.mean(similarities) - np.std(similarities)

boundaries = np.where(similarities < threshold)[0]

print("Mean similarity:", similarities.mean())
print("Std deviation:", similarities.std())
print("Boundary threshold:", threshold)
print("Number of candidate boundaries:", len(boundaries))
```

```python
segments = []

start_window = 0

for boundary in boundaries:
    end_window = boundary

    segments.append({
        "start": windows[start_window]["start"],
        "end": windows[end_window]["end"],
        "text": " ".join(
            windows[i]["text"]
            for i in range(start_window, end_window + 1)
        )
    })

    start_window = boundary + 1

# Add final segment
if start_window < len(windows):
    segments.append({
        "start": windows[start_window]["start"],
        "end": windows[-1]["end"],
        "text": " ".join(
            windows[i]["text"]
            for i in range(start_window, len(windows))
        )
    })

print("Number of segments:", len(segments))
```
